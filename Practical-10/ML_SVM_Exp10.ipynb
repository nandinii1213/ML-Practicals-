# **Implementation of Non-Linear SVM for Breast Cancer Classification**

```python
import pandas as pd
import numpy as np
import joblib

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from google.colab import files
```

```python
# Load Breast Cancer Wisconsin dataset
data = load_breast_cancer(as_frame=True)

X = data.data
y = data.target

print("Dataset loaded successfully!")
print("Original dataset shape:", X.shape)
print("Target classes:", data.target_names)

display(X.head())
```

```python
#Select six important features

features = [
    "mean radius",
    "mean texture",
    "mean perimeter",
    "mean area",
    "mean smoothness",
    "mean concavity"
]

X = X[features]

print("Selected features:")
for feature in features:
    print(feature)

print("\nDataset shape after feature selection:", X.shape)
display(X.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
# Create a pipeline with scaling and RBF-kernel SVM
model = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="rbf", C=1.0, gamma="scale"))
])

# Train the model
model.fit(X_train, y_train)

print("Non-Linear SVM model trained successfully!")
```

```python
# Predict test results
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=data.target_names
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# Select one sample from the test dataset
sample = X_test.iloc[[0]]

prediction = model.predict(sample)[0]
actual = y_test.iloc[0]

print("Input features:")
display(sample)

print("Predicted class:", data.target_names[prediction])
print("Actual class:", data.target_names[actual])
```

```python
# Save trained model
joblib.dump(model, "nonlinear_svm_model.pkl")

# Save metadata for the Flask app
metadata = {
    "features": features,
    "defaults": X.iloc[0].to_dict(),
    "target_names": list(data.target_names)
}

joblib.dump(metadata, "model_metadata.pkl")

print("Model and metadata saved successfully!")
```
