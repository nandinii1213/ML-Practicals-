**Cell 1 — Import libraries**

```python
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

**Cell 2 — Load the UCI dataset**

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00267/data_banknote_authentication.txt"

columns = [
    "Variance",
    "Skewness",
    "Curtosis",
    "Entropy",
    "Class"
]

df = pd.read_csv(url, header=None, names=columns)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

**Cell 3 — Explore the dataset**

```python
print("Dataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nClass distribution:")
print(df["Class"].value_counts())

print("\nStatistical summary:")
display(df.describe())
```

**Cell 4 — Separate features and target**

```python
X = df.drop("Class", axis=1)
y = df["Class"]

print("Features:", X.columns.tolist())
print("Target:", y.name)
```

**Cell 5 — Split the data**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

**Cell 6 — Scale the features**

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed!")
```

**Cell 7 — Train the Linear SVM model**

```python
model = SVC(kernel="linear", C=1.0)

model.fit(X_train_scaled, y_train)

print("Linear SVM model trained successfully!")
```

**Cell 8 — Evaluate the model**

```python
y_pred = model.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Genuine", "Counterfeit"]
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

**Cell 9 — Test a sample banknote**

```python
sample = pd.DataFrame(
    [[3.6216, 8.6661, -2.8073, -0.44699]],
    columns=X.columns
)

sample_scaled = scaler.transform(sample)
prediction = model.predict(sample_scaled)[0]

if prediction == 0:
    print("Prediction: Genuine Banknote")
else:
    print("Prediction: Counterfeit Banknote")
```

**Cell 10 — Save the model and scaler**

```python
joblib.dump(model, "svm_model.pkl")
joblib.dump(scaler, "scaler.pkl")

print("SVM model saved successfully!")
print("Scaler saved successfully!")
```
